```python
from ultralytics import YOLO
import os  

MODEL_PATH = "yolo11s.pt"
DATA_PATH = "/content/dataset/data.yaml"
DEVICE = "0"          
BATCH_SIZE = 32       
IMAGE_SIZE = 640      

model = YOLO(MODEL_PATH)

results = model.train(
    data=DATA_PATH,
    imgsz=IMAGE_SIZE,
    epochs=100,
    batch=BATCH_SIZE,
    device=DEVICE,
    cache=False,          
    workers=8,
    augment=True,
    mixup=0.1,
    cutmix=0.3,
    label_smoothing=0.1,
    mosaic=1.0,
    close_mosaic=10,
    lr0=0.005,
    lrf=0.0005,
    cos_lr=True,
    warmup_epochs=5,
    weight_decay=0.0005,
    translate=0.2, scale=0.7, fliplr=0.5, flipud=0.0,
    hsv_h=0.015, hsv_s=0.7, hsv_v=0.4,
    save=True,
    save_period=20,
    project="runs/train",
    name="yolo11s_final"
)

best_path = os.path.join(results.save_dir, "weights", "best.pt")
print(f"\n (Best Model): {best_path}")

best_model = YOLO(best_path)
```
